## Carregando os Dados CSV

Carregando arquivos CSV (`Divvy_Trips_2019_Q1.csv` e `Divvy_Trips_2020_Q1.csv`) em DataFrames do pandas para análise.

```python
import pandas as pd

# Caminhos dos arquivos CSV
csv_2019_q1_path = '/content/Divvy_Trips_2019_Q1.csv'
csv_2020_q1_path = '/content/Divvy_Trips_2020_Q1.csv'

# Carregar os arquivos CSV em DataFrames
try:
    df_2019_q1 = pd.read_csv(csv_2019_q1_path)
    print(f"'{csv_2019_q1_path}' carregado com sucesso.")
    display(df_2019_q1.head())
    display(df_2019_q1.info())
except FileNotFoundError:
    print(f"Erro: O arquivo '{csv_2019_q1_path}' não foi encontrado.")
except Exception as e:
    print(f"Erro ao carregar '{csv_2019_q1_path}': {e}")

try:
    df_2020_q1 = pd.read_csv(csv_2020_q1_path)
    print(f"\n'{csv_2020_q1_path}' carregado com sucesso.")
    display(df_2020_q1.head())
    display(df_2020_q1.info())
except FileNotFoundError:
    print(f"Erro: O arquivo '{csv_2020_q1_path}' não foi encontrado.")
except Exception as e:
    print(f"Erro ao carregar '{csv_2020_q1_path}': {e}")
```

## Preparação Inicial dos Dados

Converter as colunas de tempo para o formato `datetime` e fazer a verificação dos tipos de dados e informações básicas.

```python
# Converter colunas de tempo para datetime em df_2019_q1
df_2019_q1['start_time'] = pd.to_datetime(df_2019_q1['start_time'])
df_2019_q1['end_time'] = pd.to_datetime(df_2019_q1['end_time'])

# Converter colunas de tempo para datetime em df_2020_q1
df_2020_q1['started_at'] = pd.to_datetime(df_2020_q1['started_at'])
df_2020_q1['ended_at'] = pd.to_datetime(df_2020_q1['ended_at'])

print("Tipos de dados após conversão (df_2019_q1):")
display(df_2019_q1.info())
print("\nTipos de dados após conversão (df_2020_q1):")
display(df_2020_q1.info())
```

Podemos observar que os nomes das colunas e algumas de suas representações são diferentes entre os anos. Por exemplo, `tripduration` em 2019 e a duração que pode ser derivada de `started_at` e `ended_at` em 2020.

Para análise comparativa, será importante:

*   **Calcular `ride_length` para `df_2020_q1`:** Nota que em 2020, não temos uma coluna `tripduration` direta, mas podemos calculá-la a partir de `ended_at` e `started_at`.
*   **Renomear colunas para consistência:** Padronizar nomes de colunas chave para facilitar a comparação.
*   **Identificar e tratar inconsistências:** Lidar com valores ausentes ou dados inesperados.

## Padronização e Cálculo da Duração da Viagem

Padronizar a forma como a duração da viagem é registrada em ambos os DataFrames, criando uma coluna `ride_length_seconds` para ambos.

```python
# Limpar e converter 'tripduration' em df_2019_q1 para float
# A coluna 'tripduration' parece ter vírgulas como separador de milhares, o que a torna um 'object'.
df_2019_q1['tripduration'] = df_2019_q1['tripduration'].str.replace(',', '').astype(float)

# Criar 'ride_length_seconds' para df_2019_q1 (já está em segundos)
df_2019_q1['ride_length_seconds'] = df_2019_q1['tripduration']

# Calcular 'ride_length_seconds' para df_2020_q1
df_2020_q1['ride_length_seconds'] = (df_2020_q1['ended_at'] - df_2020_q1['started_at']).dt.total_seconds()

print("Verificando os tipos de dados das colunas de duração:")
print(f"df_2019_q1['ride_length_seconds'] dtype: {df_2019_q1['ride_length_seconds'].dtype}")
print(f"df_2020_q1['ride_length_seconds'] dtype: {df_2020_q1['ride_length_seconds'].dtype}")

print("\nPrimeiras linhas dos DataFrames com a nova coluna 'ride_length_seconds':")
display(df_2019_q1[['trip_id', 'start_time', 'end_time', 'tripduration', 'ride_length_seconds']].head())
display(df_2020_q1[['ride_id', 'started_at', 'ended_at', 'ride_length_seconds']].head())
```

## Limpeza e Padronização Adicional

Realizando filtragem das viagens com duração inválida (zero ou negativa) e depois renomear as colunas para que sejam consistentes entre os dois DataFrames.

```python
# Filtrar viagens com duração inválida
df_2019_q1 = df_2019_q1[df_2019_q1['ride_length_seconds'] > 0]
df_2020_q1 = df_2020_q1[df_2020_q1['ride_length_seconds'] > 0]

print(f"Número de viagens em df_2019_q1 após filtrar durações inválidas: {len(df_2019_q1)}")
print(f"Número de viagens em df_2020_q1 após filtrar durações inválidas: {len(df_2020_q1)}")

# Renomear colunas para consistência
# df_2019_q1 (original)
# trip_id -> ride_id
# start_time -> started_at
# end_time -> ended_at
# from_station_name -> start_station_name
# from_station_id -> start_station_id
# to_station_name -> end_station_name
# to_station_id -> end_station_id
# usertype -> member_casual

df_2019_q1_renamed = df_2019_q1.rename(columns={
    'trip_id': 'ride_id',
    'start_time': 'started_at',
    'end_time': 'ended_at',
    'from_station_name': 'start_station_name',
    'from_station_id': 'start_station_id',
    'to_station_name': 'end_station_name',
    'to_station_id': 'end_station_id',
    'usertype': 'member_casual'
})

# Para df_2020_q1, 'member_casual' já existe, mas vamos padronizar os valores se necessário
# (Vamos lidar com os valores de 'member_casual' em um passo posterior se houver inconsistências)

print("\nColunas de df_2019_q1 após renomeação:")
print(df_2019_q1_renamed.columns)

print("\nColunas de df_2020_q1 (para comparação):")
print(df_2020_q1.columns)
```

## Preparação Final e Unificação dos DataFrames

Garantir a consistência dos valores na coluna `member_casual` e selecionar um conjunto comum de colunas antes de unificar os DataFrames de 2019 e 2020. Também adicionaremos uma coluna para indicar o ano/trimestre, o que será crucial para a análise comparativa.

```python
# Padronizar os valores da coluna 'member_casual' em df_2019_q1_renamed
# 'Subscriber' -> 'member', 'Customer' -> 'casual'
df_2019_q1_renamed['member_casual'] = df_2019_q1_renamed['member_casual'].replace({
    'Subscriber': 'member',
    'Customer': 'casual'
})

print("Valores únicos em 'member_casual' (df_2019_q1_renamed após padronização):")
print(df_2019_q1_renamed['member_casual'].unique())
print("\nValores únicos em 'member_casual' (df_2020_q1):")
print(df_2020_q1['member_casual'].unique())

# Definir as colunas comuns para ambos os DataFrames
common_columns = [
    'ride_id',
    'started_at',
    'ended_at',
    'ride_length_seconds',
    'start_station_name',
    'start_station_id',
    'end_station_name',
    'end_station_id',
    'member_casual'
]

# Selecionar apenas as colunas comuns e criar cópias explícitas para evitar SettingWithCopyWarning
df_2019_q1_final = df_2019_q1_renamed[common_columns].copy()
df_2020_q1_final = df_2020_q1[common_columns].copy()

# Adicionar uma coluna 'quarter_year' para identificar o ano/trimestre
df_2019_q1_final['quarter_year'] = '2019_Q1'
df_2020_q1_final['quarter_year'] = '2020_Q1'

# Concatenar os dois DataFrames
df_combined = pd.concat([df_2019_q1_final, df_2020_q1_final], ignore_index=True)

print("\nInformações do DataFrame combinado:")
display(df_combined.info())
print("\nPrimeiras 5 linhas do DataFrame combinado:")
display(df_combined.head())
print("\nÚltimas 5 linhas do DataFrame combinado:")
display(df_combined.tail())
```

DataFrame `df_combined` agora está unificado, contendo os dados de viagens de Divvy para o primeiro trimestre de 2019 e 2020, com colunas consistentes e uma coluna para identificar o período, pronto para análises comparativas e exploração, que é a base para o seu estudo de caso.

## Análise Exploratória Inicial (EDA)

Realizando análises exploratórias para começar a entender os dados e identificar padrões.

```python
# 1. Estatísticas descritivas para a duração das viagens
print("Estatísticas descritivas para 'ride_length_seconds':")
display(df_combined['ride_length_seconds'].describe())

# 2. Número de viagens por tipo de usuário e por trimestre
print("\nNúmero de viagens por tipo de usuário e por trimestre:")
display(df_combined.groupby(['quarter_year', 'member_casual']).size().unstack())

# 3. Média e Mediana da duração das viagens por tipo de usuário e por trimestre
print("\nMédia da duração das viagens (em segundos) por tipo de usuário e por trimestre:")
display(df_combined.groupby(['quarter_year', 'member_casual'])['ride_length_seconds'].mean().unstack())

print("\nMediana da duração das viagens (em segundos) por tipo de usuário e por trimestre:")
display(df_combined.groupby(['quarter_year', 'member_casual'])['ride_length_seconds'].median().unstack())
```

```python
# 3. (Continuando) Mediana da duração das viagens por tipo de usuário e por trimestre
print("\nMediana da duração das viagens (em segundos) por tipo de usuário e por trimestre:")
display(df_combined.groupby(['quarter_year', 'member_casual'])['ride_length_seconds'].median().unstack())
```

Essas estatísticas iniciais podem revelar tendências importantes. É possível observar mudanças significativas no número de viagens ou na duração média das viagens entre 2019 e 2020, ou entre membros e usuários casuais.

## Tratamento de Outliers e Criação de Novas Features

Refinando coluna `ride_length_seconds` para remover outliers extremos e criar novas colunas temporais úteis para a análise.

```python
# 1. Definir um limite superior razoável para a duração das viagens (ex: 24 horas = 86400 segundos)
# Viagens muito longas podem indicar erros de dados (bicicleta não devolvida, etc.)
max_ride_length_seconds = 24 * 60 * 60  # 24 horas em segundos

print(f"Número de viagens antes de remover outliers de duração: {len(df_combined)}")
# Criar uma cópia explícita do DataFrame após a filtragem para evitar SettingWithCopyWarning
df_combined_cleaned = df_combined[df_combined['ride_length_seconds'] <= max_ride_length_seconds].copy()
print(f"Número de viagens após remover outliers de duração (<= 24h): {len(df_combined_cleaned)}")

# 2. Criar novas colunas temporais
df_combined_cleaned['start_day_of_week'] = df_combined_cleaned['started_at'].dt.day_name()
df_combined_cleaned['start_hour'] = df_combined_cleaned['started_at'].dt.hour
df_combined_cleaned['start_month'] = df_combined_cleaned['started_at'].dt.month_name()

print("\nDataFrame com outliers removidos e novas colunas temporais (primeiras 5 linhas):")
display(df_combined_cleaned.head())
```

Agora, vamos analisar como o uso da bicicleta varia por dia da semana, hora do dia e mês, e comparar esses padrões entre os anos e os tipos de usuário.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calcular o total de viagens por tipo de usuário
rides_by_user_type = df_combined_cleaned.groupby('member_casual').size().reset_index(name='total_rides')

# Criar o gráfico de barras
fig = plt.figure(figsize=(8, 6))
sns.barplot(x='member_casual', y='total_rides', data=rides_by_user_type, palette='viridis')
plt.title('Total de Viagens por Tipo de Usuário (Membro vs. Casual)')
plt.xlabel('Tipo de Usuário')
plt.ylabel('Total de Viagens')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calcular a duração média das viagens por tipo de usuário e dia da semana
average_ride_duration_by_day = df_combined_cleaned.groupby(['member_casual', 'start_day_of_week'])['ride_length_seconds'].mean().reset_index()

# Para uma melhor visualização, podemos ordenar os dias da semana
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
average_ride_duration_by_day['start_day_of_week'] = pd.Categorical(average_ride_duration_by_day['start_day_of_week'], categories=day_order, ordered=True)
average_ride_duration_by_day = average_ride_duration_by_day.sort_values('start_day_of_week')

print("Duração média das viagens (em segundos) por tipo de usuário e dia da semana:")
display(average_ride_duration_by_day)

# Opcional: Criar um gráfico de barras para visualizar esses dados
fig = plt.figure(figsize=(12, 7))
sns.barplot(x='start_day_of_week', y='ride_length_seconds', hue='member_casual', data=average_ride_duration_by_day, palette='Paired')
plt.title('Duração Média das Viagens por Dia da Semana e Tipo de Usuário')
plt.xlabel('Dia da Semana')
plt.ylabel('Duração Média da Viagem (segundos)')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend(title='Tipo de Usuário')
plt.tight_layout()
plt.show()
```

```python
# Adicionar uma coluna 'is_weekend' para facilitar a análise
weekend_days = ['Saturday', 'Sunday']
df_combined_cleaned['is_weekend'] = df_combined_cleaned['start_day_of_week'].isin(weekend_days)

# Calcular estatísticas descritivas para a duração das viagens nos finais de semana vs dias de semana, por tipo de usuário
ride_duration_weekend_analysis = df_combined_cleaned.groupby(['member_casual', 'is_weekend'])['ride_length_seconds'].describe().reset_index()

print("Estatísticas descritivas da duração das viagens (em segundos) por tipo de usuário e se é final de semana:")
display(ride_duration_weekend_analysis)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Criar um box plot para visualizar a duração das viagens por tipo de usuário e se é final de semana
fig = plt.figure(figsize=(10, 7))
sns.boxplot(x='is_weekend', y='ride_length_seconds', hue='member_casual', data=df_combined_cleaned, palette='coolwarm')
plt.title('Distribuição da Duração das Viagens (segundos) por Final de Semana e Tipo de Usuário')
plt.xlabel('É Final de Semana (False = Dia de Semana, True = Final de Semana)')
plt.ylabel('Duração da Viagem (segundos)')
plt.xticks([0, 1], ['Dia de Semana', 'Final de Semana'])
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend(title='Tipo de Usuário')
plt.tight_layout()
plt.show()
```

## Principais descobertas

Com base na análise que realizamos até agora com os dados do Divvy Bike-Share de Chicago para o primeiro trimestre de 2019 e 2020, as principais descobertas para o estudo de caso são:

1. Crescimento Generalizado, Impulsionado por Usuários Casuais: Observamos um aumento geral no número de viagens de 2019 para 2020 no primeiro trimestre. Este crescimento foi particularmente notável entre os usuários casuais, que mostraram um incremento mais significativo no volume de viagens em comparação com os membros.

2. Diferença Acentuada na Duração das Viagens:

  - Membros tendem a ter durações de viagem mais curtas e consistentes, sugerindo um uso mais focado em deslocamentos (casa-trabalho, por exemplo).
  - Usuários Casuais realizam viagens significativamente mais longas. A média da duração das viagens para usuários casuais é consideravelmente maior do que para os membros.

3. Padrões de Uso nos Finais de Semana:

  - A diferença na duração das viagens é ainda mais pronunciada nos finais de semana. Usuários casuais fazem viagens substancialmente mais longas aos sábados e domingos em comparação com os dias de semana.
  - Membros também apresentam um leve aumento na duração das viagens nos finais de semana, mas essa variação é bem menor do que a observada nos usuários casuais. Isso reforça a hipótese de que usuários casuais utilizam as bicicletas para lazer e atividades recreativas, enquanto membros as usam para necessidades mais rotineiras, mesmo nos finais de semana.
  - Consistência dos Membros: A média e a mediana da duração das viagens para membros são consistentemente mais baixas e menos variáveis, indicando um comportamento de uso mais preditivo e estruturado.

4. Implicações:

Essas descobertas sugerem que a empresa de compartilhamento de bicicletas pode ter diferentes estratégias para atrair e reter cada tipo de usuário. Por exemplo, focar em converter usuários casuais que fazem viagens longas nos finais de semana em membros pode ser uma oportunidade. Além disso, entender os padrões de uso em dias e horários específicos pode ajudar na otimização da distribuição e manutenção das bicicletas.
